## Exercise 1
Criminologists have long debated whether there is a relationship between weather conditions and the number of crimes committed. In 1988, an article “Is there a
Season for Homicide?” (Criminology, 1988: 287-296) classified 1361 homicides by season, resulting
in the data below.  
Is there an association between season and the number of murders committed?


```python
dfmurders = pd.DataFrame(data={
    'season': ['winter', 'spring','summer','autumn'],
    'number_of_murders': [328, 334, 372, 327]
})
dfmurders
```

> stap 1:  
> $H_0$: Er is geen verband tussen de seizoenen en het aantal moorden  
> $H_1$: Er is een verband tussen de seizoenen en het aantal moorden

```python
# Stap 2 significance level
alpha = 0.05
# stap 3 -> teststatistiek = goodness-of-fit test
# stap 4 -> p + g berekenen
observed =   dfmurders["number_of_murders"]
expected_p = np.array([.25, .25, .25, .25])
alpha = 0.05               # Significance level
n = sum(observed)          # Sample size
k = len(observed)          # Number of categories
dof = k - 1                # Degrees of freedom
expected = expected_p * n  # Expected absolute frequencies in the sample
g = stats.chi2.isf(alpha, df=dof)  # Critical value

# Goodness-of-fit-test in Python:
chi2, p = stats.chisquare(f_obs=observed, f_exp=expected)

print("Significance level  ⍺ = %.2f" % alpha)
print("Sample size         n = %d" % n)
print("k = %d; df = %d" % (k, dof))
print("Chi-squared        χ² = %.4f" % chi2)
print("Critical value      g = %.4f" % g)
print("p-value             p = %.4f" % p)
```

> stap 5 -> conclusie  
> $p$ $\gt$ $\alpha$ en $g$ $\gt$ $\chi$²  
> Er is niet voldoende aanwijzing om $H_0$ te verwerpen -> geen verband

## Exercise 2
A random sample of individuals was taken, and each individual was classified according to their level of education and their sense of spirituality.
Is there an association between the level of education and the sense of spirituality?

```python
dfspirituality = pd.DataFrame(data={
    'sense_of_spirituality': ['lot', 'moderate','little','none'],
    'alfa_scientist': [56, 162, 198, 211],
    'beta_scientist': [56, 223, 243, 239],
    'not_scientist': [109, 164, 74, 28]
})
dfspirituality.head()
```

> stap 1:  
> $H_0$: Er is geen verband tussen het type opleiding en het geloof in spiritualiteit  
> $H_1$: Er is een verband tussen het type opleiding en het geloof in spiritualiteit

```python
# Stap 2 significance level
alpha = 0.05
# stap 3 -> teststatistiek = chi2 test
# stap 4 -> p + g berekenen
observed = dfspirituality.drop('sense_of_spirituality', axis=1)

chi2, p, dof, expected = stats.chi2_contingency(observed)
g = stats.chi2.isf(alpha, df = dof)
print(f"χ²                ≈ {chi2:.3f}")
print(f"Degrees of freedom: {dof}")
print(f"Critical value    ≃ {g:.3f}")
print(f"p-value           : {p:.4f}")
```

> stap 5 -> conclusie  
> $p$ $\lt\lt$ $\alpha$ en $g$ $\lt\lt$ $\chi$²  
> Er is voldoende aanwijzing om $H_0$ te verwerpen -> verband

```python
stats.contingency.association(observed, method='cramer')
```

## Exercise 4

In an experiment, 150 patients were divided into three groups of 45, 45 and 60 patients each. Two groups were given a new drug in different doses and the third group was given a placebo. The way the patients felt after using the drug are shown in the data frame below. Is there evidence at the 5% significance level that there is a difference in the condition between the three groups of patients?



```python
dfexperiment = pd.DataFrame(data={
    'state': ['better', 'equal', 'worse'],
    'placebo': [16, 20, 9],
    'half_dosis': [17, 18, 10],
    'full_dosis': [26, 20, 14]
}).set_index('state')
dfexperiment.head()
```

> stap 1:  
> $H_0$: Er is geen verband tussen de dosis en hun gevoel na de drug  
> $H_1$: Er is een verband tussen de dosis en hun gevoel na de drug

```python
# Stap 2 significance level
alpha = 0.05
# stap 3 -> teststatistiek = chi2 test
# stap 4 -> p + g berekenen
observed = dfexperiment

chi2, p, dof, expected = stats.chi2_contingency(observed)
g = stats.chi2.isf(alpha, df = dof)
print(f"χ²                ≈ {chi2:.3f}")
print(f"Degrees of freedom: {dof}")
print(f"Critical value    ≃ {g:.3f}")
print(f"p-value           : {p:.4f}")
```

> stap 5:  
> $p$ $\gt\gt$ $\alpha$ en $g$ $\gt\gt$ $\chi$²  
> Er is niet voldoende aanwijzing om $H_0$ te verwerpen -> correcte sample
