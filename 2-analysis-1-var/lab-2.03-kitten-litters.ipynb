## Exercise 3 - Kitten litters

A vet keeps a record of the number of kittens in litters produced by cats in his practice. The code block below contains the data as a frequency table.

- First put the data into one list of individual observations: [1, 1, 2, 2, ... 5, 5, 6, 6, 6, 6, 7, 7, 8]  
- Visualise the data set using all suitable chart types that you know of.
- Calculate:
    - the median size of a litter
    - the mean size of a litter
    - the lower quartile
    - the upper quartile

```python
kittens = pd.DataFrame.from_dict({
    'Number in litter': [1, 2, 3,  4, 5, 6, 7, 8],
    'Frequency':        [2, 4, 7, 11, 8, 4, 2, 1]})
kittens
```

```python
# Slow:
list = []
for (index, i) in enumerate(kittens['Number in litter']):
    for j in range(kittens['Frequency'][index]):
        list.append(i)
print(list)

# Little better:
better_list = []
for number, freq in zip(kittens['Number in litter'], kittens['Frequency']):
    better_list.extend([number] * freq)
print(better_list)

# Best:
kitten_list = kittens['Number in litter'].repeat(kittens['Frequency']).tolist()
print(kitten_list)
```

```python
sns.barplot(data=kittens, x='Number in litter', y='Frequency');
```

```python
print(f'Median: {np.median(kitten_list)}')
print(f'Median: {np.mean(kitten_list)}')
print(f'Median: {np.quantile(kitten_list, .25)}')
print(f'Median: {np.quantile(kitten_list, .75)}')
```
