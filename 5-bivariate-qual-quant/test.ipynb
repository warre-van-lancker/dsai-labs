# Question

If you flip a coin you either get heads (H) or tails (T).  
What are the chances of the following events happening?  
You flip a coin until you reach one of the following situations:  
 - The last three throws were `HTT`
 - The last three throws were `TTH`

```python
import random

def simulate_coin_game(trials=1000, seq1='HTT', seq2='TTH'):
    choices = ['H', 'T']
    first, second = 0, 0
    
    for _ in range(trials):
        window = "".join(random.choices(choices, k=3))
        
        while True:
            if window == seq1:
                first += 1
                break
            elif window == seq2:
                second += 1
                break
                
            window = window[1:] + random.choice(choices)
            
    print(f"Player 1 ({seq1}): {first} wins")
    print(f"Player 2 ({seq2}): {second} wins")

simulate_coin_game(1000)
```
