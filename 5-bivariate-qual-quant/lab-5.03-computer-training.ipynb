$$\Delta = \frac{|\overline{x}_1 - \overline{x}_2|}{s_1}$$

```python
computer_skills = pd.read_csv('https://raw.githubusercontent.com/HoGentTIN/dsai-labs/main/data/Computer%20Skills.csv', delimiter=';')
computer_skills.head()
```

```python
sns.boxplot(data=computer_skills[['Score Before', 'Score After']])
```

> stap 1:  
> $H_0$: er zal geen significant verschil zijn tussen before and after  
> $H_1$: er zal wel een significant verschil zijn tussen before and after

```python
# stap 2 -> significance level
alpha = 0.05

# stap 3 -> test statistiek (paired 2 sample t test)
# stap 4 -> bereken p
t, p =stats.ttest_rel(computer_skills['Score Before'], computer_skills['Score After'], alternative='less')

print(f"t-value   :   {t:.5f}")
print(f"p-value   :   {p:.5f}")
```

> stap 5: 
> $p$ $\lt$ $\alpha$ en $g$ $\lt$ $\chi$²  
> Er is voldoende aanwijzing om $H_0$ te verwerpen -> significant verschil tussen before and after

```python
def cohen_d(a, b):
    na = len(a)
    nb = len(b)
    pooled_sd = np.sqrt( ((na-1) * np.var(a, ddof=1) +
                          (nb-1) * np.var(b, ddof=1)) / (na + nb - 2) )
    return (np.mean(b) - np.mean(a)) / pooled_sd
```

```python
# Effect size for the clinical trial example
d = cohen_d(computer_skills['Score Before'], computer_skills['Score After'])
print(f"Cohen's value: {d:.5f}")
```
