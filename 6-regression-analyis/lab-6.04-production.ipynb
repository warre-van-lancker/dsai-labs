```python
df = pd.read_csv('https://raw.githubusercontent.com/HoGentTIN/dsai-labs/main/data/production.csv', delimiter=";")
df.head(20)
```

```python
sns.regplot(data=df, x='unit_produced', y='production_cost')
```

2.1 -> negative relation, because regression line goes downward (x goes up then y goes down)  
2.2 -> strong relation, because the dots are not far from the regression line  
2.3 -> between -1 and -0.5 -> because there is a strong relation (0.5, 1) and it is negative relation (-)

```python
# Calculate Beta_0 and Beta_1
b1, b0 = np.polyfit(x=df.unit_produced, y=df.production_cost, deg=1)
print(f"Beta_0 = {b0:.5f}")
print(f"Beta_1 = {b1:.5f}")
print(f"Regression line: ŷ = {b0} + {b1}x")
```

```python
# 3
x = 400
print(f"ŷ = {b0} + {b1} * {x} = {b0 + b1 * x}")
# Answer: 46.50515463917526
```

```python
# 4
x = 0
print(f"ŷ = {b0} + {b1} * {x} = {b0 + b1 * x}")
# Answer: 81.3917525773196
```

```python
x = 1000
total = 0
for item in range(1, x + 1):
    total += b0 + b1 * item
print(f"Total price: {total}")
```

```python
# Shorter
total = sum([b0 + b1 * x for x in range(1, 1001)])
print(f"Total price: {total}")
```
